**Question:** Using a time interval, how could you observe trends in the companies that became unicorns in one year?


I would group the companies that joined in 2021 by a time interval, such as week or month, and count how many reached unicorn status in each interval. Since Date Joined is a datetime column, I can use dt.to_period("W") to group by week (or "M" for month). Comparing the counts across intervals shows when unicorn status was gained most often during the year, and whether the pace rose, fell, or stayed steady.

### Observe trends over time

Implement the structuring approach that you have identified to observe trends over time in the companies that became unicorns for the year that interests you.

```python
# Step 1. Insert a column with the time interval (week) each company joined in.
companies_2021 = companies_2021.copy()
companies_2021["Week Joined"] = companies_2021["Date Joined"].dt.strftime("%Y-W%V")

# Step 2. Group by the time interval and count companies that joined per week.
companies_by_week_2021 = (
    companies_2021.groupby("Week Joined")["Company"]
    .count()
    .reset_index()
    .rename(columns={"Company": "Company Count"})
)

# Display the first few rows of the new DataFrame to confirm that it was created
companies_by_week_2021.head()
```

**Question:** How would you structure the data to observe trends in the average valuation of companies from 2020 to 2021?  

I would filter the data to companies that joined in 2020 and 2021, convert Valuation to numbers, then group by a time interval such as quarter and take the average valuation for each interval. This gives one row per quarter, so I can compare how the average valuation rose or fell over the two years.

### Compare trends over time

Implement the structuring approach that you have identified in order to compare trends over time in the average valuation of companies that became unicorns in the year you selected above and in another year of your choice. Keep in mind the data type of the `Valuation` column and what the values in that column contain currently.

```python
# Step 1. Filter by the additional year (2020) to create a subset of companies that joined that year.
companies_2020 = companies[companies["Date Joined"].dt.year == 2020].copy()

# Step 2. Concatenate that new subset with the 2021 subset defined previously.
companies_2020_2021 = pd.concat([companies_2020, companies_2021.drop(columns="Week Joined")], ignore_index=True)

# Step 3. Add a column that contains the time interval (quarter) each data point belongs to.
companies_2020_2021["Quarter Joined"] = companies_2020_2021["Date Joined"].dt.to_period("Q").dt.strftime("%Y-Q%q")

# Step 4. Transform the `Valuation` column into numbers (in billions).
companies_2020_2021["Valuation Num"] = companies_2020_2021["Valuation"].str.strip("$B").astype(int)

# Step 5. Group by the time interval and compute the average valuation per quarter.
companies_by_quarter_2020_2021 = (
    companies_2020_2021.groupby("Quarter Joined")["Valuation Num"]
    .mean()
    .reset_index()
    .rename(columns={"Valuation Num": "Average Valuation (B)"})
)

# Display the first few rows of the new DataFrame to confirm that it was created.
companies_by_quarter_2020_2021.head()
```

### Visualize the time it took companies to become unicorns

Using the `companies` dataset, create a box plot to visualize the distribution of how long it took companies to become unicorns, with respect to the month they joined. 

```python
# Define a list that contains months in chronological order.
month_order = ["January", "February", "March", "April", "May", "June",
               "July", "August", "September", "October", "November", "December"]

# Print out the list to confirm it is correct.
print(month_order)
```

```python
# Create the box plot to visualize the distribution of how long it took companies to become unicorns, with respect to the month they joined.
# Make sure the x-axis goes in chronological order by month, using the list you defined previously.
# Plot the data from the `companies` DataFrame.
sns.boxplot(x=companies["Month Joined"],
            y=companies["Years To Join"],
            order=month_order)

# Set the title of the plot.
plt.title("Distribution of Time Taken to Become a Unicorn by Month Joined")

# Rotate labels on the x-axis as a way to avoid overlap in the positions of the text.
plt.xticks(rotation=45, horizontalalignment="right")

# Display the plot.
plt.show()
```
